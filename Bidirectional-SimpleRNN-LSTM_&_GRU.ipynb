```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,Bidirectional, SimpleRNN, LSTM, GRU, Dense, Input
```

```python
num_words=10000
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words=num_words)
```

```python
maxlen = 100
x_train = pad_sequences(x_train,maxlen=maxlen,padding='post', truncating = 'post')
x_test = pad_sequences(x_test, maxlen=maxlen,padding='post', truncating = 'post')
```

```python
embedding_dim=32
```

```python
model = Sequential([
    Input(shape=(100,)),
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=maxlen),
    Bidirectional(SimpleRNN(5)),
    Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history=model.fit(x_train,y_train,epochs=5, batch_size=32,validation_split=0.2)
```

```python
LSTM_model=Sequential([
    Input(shape=(100,)),
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=maxlen),
    Bidirectional(LSTM(5)),
    Dense(1,activation='sigmoid', )
])
model.summary()
```

```python
LSTM_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
LSTM_history=LSTM_model.fit(x_train,y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
GRU_model=Sequential([
    Input(shape=(100,)),
    Embedding(input_dim=num_words, output_dim=embedding_dim,input_length=maxlen),
    Bidirectional(GRU(5)),
    Dense(1,activation='sigmoid')
])
GRU_model.summary()
```

```python
GRU_model.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])
```

```python
GRU_history=GRU_model.fit(x_train,y_train, epochs=5,batch_size=32, validation_split=0.2)
```
